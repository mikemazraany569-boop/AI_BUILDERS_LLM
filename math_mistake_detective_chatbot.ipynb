```python
import ollama

# ============================================
# 1. SYSTEM PROMPT
# ============================================

system_prompt = """
You are Math Mistake Detective, an AI mathematics tutor designed to help
high-school and first-year university students understand and correct
mistakes in their mathematical reasoning.

Your main purpose is to analyze a student's attempted solution, identify
the first incorrect step, explain why it is wrong, and guide the student
toward the correct reasoning.

When a student provides a mathematics problem and their attempted solution:

1. Break the solution into clear steps.
2. Check each step in order.
3. Identify the FIRST incorrect step.
4. Explain why that step is incorrect.
5. Classify the mistake when possible:
   - arithmetic error
   - sign error
   - algebraic error
   - formula misuse
   - conceptual misunderstanding
   - logical error
   - unit/conversion error
6. Give the student a useful hint.
7. Provide the complete solution when requested.
8. Encourage the student to try a similar problem.

Communication style:
- Be patient, encouraging, and respectful.
- Use simple explanations appropriate for students.
- Explain the reasoning rather than only giving the answer.
- Do not make the student feel embarrassed about mistakes.
- Use mathematical notation and examples when useful.

If the student's solution is correct, clearly tell them that it is correct.

If the problem is ambiguous or incomplete, ask for clarification.

Your goal is not simply to solve mathematics problems.
Your goal is to help students understand why their reasoning is
correct or incorrect and how to improve it.
"""


# ============================================
# 2. GENERATION PARAMETERS
# ============================================

# These values can be changed during the conversation
# using the "settings" command.

temperature = 0.3
num_predict = 500


# ============================================
# 3. CONVERSATION HISTORY
# ============================================

# The system prompt is kept throughout the conversation.
# User and assistant messages are added below so the chatbot
# can understand follow-up questions.

messages = [
    {
        "role": "system",
        "content": system_prompt
    }
]


# ============================================
# 4. DISPLAY CHATBOT INFORMATION
# ============================================

print("=" * 65)
print("                 MATH MISTAKE DETECTIVE")
print("=" * 65)

print("Model: llama3.2")
print(f"Temperature: {temperature}")
print(f"Maximum output tokens: {num_predict}")

print("\nCommands:")
print("  settings  - change temperature or maximum output tokens")
print("  clear     - clear the conversation history")
print("  exit      - quit the chatbot")
print("=" * 65)
print()


# ============================================
# 5. CHAT LOOP
# ============================================

while True:

    user_input = input("You: ").strip()

    # --------------------------------------------
    # Exit chatbot
    # --------------------------------------------

    if user_input.lower() in ["exit", "quit"]:
        print("\nGoodbye!")
        break


    # --------------------------------------------
    # Change generation parameters
    # --------------------------------------------

    if user_input.lower() == "settings":

        print("\nCurrent generation settings:")
        print(f"Temperature: {temperature}")
        print(f"Maximum output tokens: {num_predict}")

        try:

            # Temperature
            new_temperature = input(
                "\nEnter temperature "
                "(0.0 - 2.0, Enter to keep current): "
            ).strip()

            if new_temperature:
                new_temperature = float(new_temperature)

                if 0.0 <= new_temperature <= 2.0:
                    temperature = new_temperature
                else:
                    print("Temperature must be between 0.0 and 2.0.")
                    continue


            # Maximum output tokens
            new_tokens = input(
                "Enter maximum output tokens "
                "(Enter to keep current): "
            ).strip()

            if new_tokens:
                new_tokens = int(new_tokens)

                if new_tokens > 0:
                    num_predict = new_tokens
                else:
                    print("Maximum output tokens must be greater than 0.")
                    continue


            print("\nSettings updated successfully.")
            print(f"Temperature: {temperature}")
            print(f"Maximum output tokens: {num_predict}\n")

        except ValueError:

            print("\nInvalid value.")
            print("Settings were not changed.\n")

        continue


    # --------------------------------------------
    # Clear conversation history
    # --------------------------------------------

    if user_input.lower() == "clear":

        messages = [
            {
                "role": "system",
                "content": system_prompt
            }
        ]

        print("\nConversation history cleared.\n")
        continue


    # ============================================
    # 6. ADD USER MESSAGE TO HISTORY
    # ============================================

    messages.append({
        "role": "user",
        "content": user_input
    })


    # ============================================
    # 7. STREAM RESPONSE
    # ============================================

    print("\nBot: ", end="", flush=True)

    try:

        response = ollama.chat(
            model="llama3.2",
            messages=messages,
            stream=True,
            options={
                "temperature": temperature,
                "num_predict": num_predict
            }
        )

        full_response = ""

        # Display each generated chunk immediately
        for chunk in response:

            text = chunk["message"]["content"]

            print(text, end="", flush=True)

            full_response += text


        print("\n")


        # ========================================
        # 8. SAVE ASSISTANT RESPONSE
        # ========================================

        messages.append({
            "role": "assistant",
            "content": full_response
        })


    except Exception as e:

        print("\n\nAn error occurred:")
        print(e)

        # Remove the user message if the API call failed.
        # This prevents an unsuccessful request from remaining
        # in the conversation history.

        messages.pop()

        print()

```
